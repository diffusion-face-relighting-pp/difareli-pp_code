```python
import numpy as np
import glob, os
import pandas as pd

fs = glob.glob('./*.txt')
df = []
for f in fs:
    df.append(pd.read_csv(f, delimiter='\t', header=None, names=['img_name', 'has_shadow']))

df = pd.concat(df)
df['has_shadow'] = df['has_shadow'].astype(int)

shadow = df.loc[df['has_shadow'] == 1]
no_shadow = df.loc[df['has_shadow'] == 0]
print("with shadow : ", len(shadow))
print("with no shadow : ", len(no_shadow))
```

```python
p_fs = '/data/mint/DPM_Dataset/ffhq_256_with_anno/params/train/ffhq-train-shadow-anno.txt'
shadow_logtis_df = pd.read_csv(p_fs, delimiter=' ', header=None, names=['img_name', 'shadow_logits'])

no_shadow_l = list(no_shadow['img_name'].values)
no_shadow_arr = []
for name in no_shadow_l:
    no_shadow_arr.append(shadow_logtis_df.loc[shadow_logtis_df['img_name']==name]['shadow_logits'].values[0])
    
shadow_l = list(shadow['img_name'].values)
shadow_arr = []
for name in shadow_l:
    shadow_arr.append(shadow_logtis_df.loc[shadow_logtis_df['img_name']==name]['shadow_logits'].values[0])
# shadow_logtis_df[shadow_logtis_df['img_name'].isin(no_shadow_l)]
# shadow_logtis_dfuery(f"img_name in {no_shadow_l}")

# shadow_l = list(shadow['img_name'])
# shadow_logtis_df[shadow_logtis_df['img_name'].isin(shadow_l)]
```
